### Quantile

`quantile()` tells us the value at a specific position in the sorted data.

- `quantile(0)` → Minimum value (0th percentile)
- `quantile(0.25)` → 25th percentile (Q1)
- `quantile(0.50)` → 50th percentile = **Median**
- `quantile(0.75)` → 75th percentile (Q3)
- `quantile(1)` → Maximum value (100th percentile)

For example, if we use `delivery_days`:

- `0` → shortest delivery time
- `0.50` → median delivery time
- `1` → longest delivery time

**Important:** The number inside `quantile()` represents the position/percentile of the data, from 0 to 1.

```python
orders["delivery_days"].quantile()
```
